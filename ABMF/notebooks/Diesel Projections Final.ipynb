```python
import sqlalchemy
import pandas as pd
import numpy as np
from google.colab import userdata
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
from matplotlib.colors import LinearSegmentedColormap
import matplotlib.ticker as mtick
```

```python
DB_USER = userdata.get('DB_USER_2')
DB_PASSWORD = userdata.get('DB_PASSWORD_2')
DB_HOST = userdata.get('DB_HOST_2')
DB_NAME = userdata.get('DB_NAME_2')
DB_PORT = "5432"
```

```python
connection_string = f"postgresql+psycopg2://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}"
engine = sqlalchemy.create_engine(connection_string)
```

```python
pd.set_option('display.max_columns', None)
```

```python
# Earliest dates

earliest_date = """
SELECT time
FROM public.abmf_table_2024
ORDER BY time ASC
LIMIT 1;
"""

earliest_date_df = pd.read_sql(earliest_date, engine)
```

```python
# Latest date

latest_date = """
SELECT time
FROM public.abmf_table_2024
ORDER BY time DESC
LIMIT 1;
"""

latest_date_df = pd.read_sql(latest_date, engine)
```
