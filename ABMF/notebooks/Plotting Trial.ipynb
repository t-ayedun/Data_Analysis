```python
import matplotlib.pyplot as plt
import numpy as np

# =========================================================
# CONFIGURATION: Change these values to fit your needs
# =========================================================

# 1. Define your range (from_value to to_value)
start_value = 0
end_value = 64
num_points = 65  # Higher number means a smoother curve

# 2. Generate the x values
x = np.linspace(start_value, end_value, num_points)

# 3. Define your equation here
# Use ** for exponents (e.g., x**2 for x squared)
y = 0.717/x + 0.272

# =========================================================
# PLOTTING
# =========================================================

# Create the figure
plt.figure(figsize=(8, 5))

# Plot the equation
plt.plot(x, y, label="y = x^2 - 3x + 2", color="blue", linewidth=2)

# Customize the chart
plt.title("Plot of the Equation", fontsize=14, fontweight="bold")
plt.xlabel("X Axis", fontsize=12)
plt.ylabel("Y Axis", fontsize=12)

# Add reference lines for the axes
plt.axhline(0, color="black", linewidth=0.8, linestyle="--")
plt.axvline(0, color="black", linewidth=0.8, linestyle="--")

# Add a grid and a legend
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend()

# Display the plot
plt.show()
```
